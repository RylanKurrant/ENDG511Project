```python
import pandas as pd

# Load the CSV file
df = pd.read_csv("ENDG511 Project Temps and Humidity.csv")


# Show the first few rows
print(df.head())
```

```python
predicted_value = 'Aloe Vera'
matching_row = df[df.iloc[:, 0] == predicted_value]
print(matching_row)
```

```python
# Select specific columns (e.g., column 2 and 3)
selected_columns = matching_row.iloc[:, [3, 6]]
print(selected_columns)
```

```python
# Extracting temperature and humidty values

temp = selected_columns.iloc[0, 0]
hum = selected_columns.iloc[0, 1]

print('The Ideal Temperature is: ', temp)
print('The Ideal Humidity is: ', hum)
```

```python
# All combined
import pandas as pd

predicted_value = 'African Violet'

# Load the CSV file
df = pd.read_csv("ENDG511 Project Temps and Humidity.csv")

# Find Matching Row
matching_row = df[df.iloc[:, 0] == predicted_value]
print(matching_row)

# Extract Ideal Temp and Hum Data
selected_columns = matching_row.iloc[:, [3, 6]]
temp = selected_columns.iloc[0, 0]
hum = selected_columns.iloc[0, 1]
print('The Ideal Temperature for ', predicted_value,' is: ', temp)
print('The Ideal Humidity for ', predicted_value,' is: ', hum)
```
