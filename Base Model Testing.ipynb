```python
from fastai.vision.all import *
```

```python
#Import Model
model_path = 'fine_tuned_plant_classifier_model_v1_base.pkl'
learn = load_learner(model_path)
```

```python
# Load the DataLoaders
with open('v1_dls.pkl', 'rb') as f:
    learn.dls = pickle.load(f)
```

```python
print(len(learn.dls.valid_ds))
```

```python
import torch

# Save only the model's state_dict (weights)
torch.save(learn.model.state_dict(), 'base_model.pth')
```

```python
# Confusion Matrix
from fastai.interpret import ClassificationInterpretation

interp = ClassificationInterpretation.from_learner(learn)
interp.plot_confusion_matrix(figsize=(8,8))
```

```python
import matplotlib.pyplot as plt

# Load image
img = Image.open('K_image.jpg')

# Get prediction
pred_class, pred_idx, probs = learn.predict('K_image.jpg')

# Display image with predicted class
plt.imshow(img)
plt.axis('off')  # Hide axis
plt.title(f'Predicted: {pred_class}', fontsize=14, color='black')  # Show prediction
plt.show()
```
